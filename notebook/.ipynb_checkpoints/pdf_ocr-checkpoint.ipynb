```python
import fitz
import pytesseract
import pandas as pd
import re
import shutil

from pathlib import Path
from PIL import Image

print("Libraries imported successfully!")
print("Tesseract:", shutil.which("tesseract"))
```

```python
DATASET_PATH = Path(
    "/home/student/Desktop/python/pdf_document_classification/datasets"
)

categories = ["invoice", "receipt", "bills"]

for category in categories:
    folder = DATASET_PATH / category

    if folder.exists():
        print(category, ":", len([
            f for f in folder.iterdir()
            if f.is_file()
        ]), "files")
    else:
        print("Folder not found:", folder)
```

```python
def extract_text_from_pdf(file_path):
    extracted_pages = []
    methods = []

    with fitz.open(file_path) as doc:
        for page in doc:
            # First, try normal PDF text extraction
            text = page.get_text(sort=True)

            if len(text.strip()) < 20:
                # Use OCR if the page has little extractable text
                text_page = page.get_textpage_ocr(
                    language="eng",
                    dpi=300
                )

                text = page.get_text(
                    textpage=text_page,
                    sort=True
                )

                methods.append("OCR")
            else:
                methods.append("PyMuPDF")

            extracted_pages.append(text)

    return "\n".join(extracted_pages), methods


def extract_text_from_image(file_path):
    with Image.open(file_path) as image:
        text = pytesseract.image_to_string(
            image,
            lang="eng"
        )

    return text, ["Tesseract OCR"]


print("OCR extraction functions created successfully!")
```

```python
records = []

supported_extensions = {
    ".pdf", ".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp"
}

for category in categories:
    folder = DATASET_PATH / category

    if not folder.exists():
        print("Skipping missing folder:", folder)
        continue

    for file_path in folder.rglob("*"):

        if not file_path.is_file():
            continue

        if file_path.suffix.lower() not in supported_extensions:
            continue

        try:
            if file_path.suffix.lower() == ".pdf":
                text, methods = extract_text_from_pdf(file_path)
            else:
                text, methods = extract_text_from_image(file_path)

            status = (
                "success"
                if text.strip()
                else "empty_text"
            )

            records.append({
                "filename": file_path.name,
                "file_path": str(file_path),
                "category": category,
                "file_type": file_path.suffix.lower(),
                "extracted_text": text,
                "extraction_method": ", ".join(sorted(set(methods))),
                "status": status,
                "error": ""
            })

        except Exception as e:
            records.append({
                "filename": file_path.name,
                "file_path": str(file_path),
                "category": category,
                "file_type": file_path.suffix.lower(),
                "extracted_text": "",
                "extraction_method": "failed",
                "status": "failed",
                "error": str(e)
            })

        if len(records) % 50 == 0:
            print("Processed", len(records), "documents")

df = pd.DataFrame(records)

print("\nOCR processing completed!")
print("Total documents:", len(df))
print("\nCategory counts:")
print(df["category"].value_counts())

print("\nExtraction status:")
print(df["status"].value_counts())
```

```python
import pandas as pd
import re
from pathlib import Path

BASE_DIR = Path(
    "/home/student/Desktop/python/pdf_document_classification"
)

df = pd.read_csv(BASE_DIR / "csv_file.csv")

print("Total documents:", len(df))
print("\nColumns:", df.columns.tolist())

# Keep only successful extractions and valid categories
df_ml = df[
    (df["status"] == "success") &
    (df["category"].isin(["invoice", "receipt", "bills"])) &
    (df["extracted_text"].notna())
].copy()

# Remove empty text
df_ml["extracted_text"] = (
    df_ml["extracted_text"].astype(str).str.strip()
)

df_ml = df_ml[
    df_ml["extracted_text"] != ""
].copy()

# Clean extracted text
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df_ml["cleaned_text"] = df_ml["extracted_text"].apply(clean_text)

df_ml = df_ml[
    df_ml["cleaned_text"].str.strip() != ""
].copy()

print("\nUsable documents:", len(df_ml))
print("\nCategory distribution:")
print(df_ml["category"].value_counts())
```

```python
import matplotlib.pyplot as plt

category_counts = df_ml["category"].value_counts()

print("Category counts:")
print(category_counts)

print("\nCategory percentages:")
print(
    (category_counts / category_counts.sum() * 100).round(2)
)

category_counts.plot(kind="bar")

plt.title("Document Distribution Before Balancing")
plt.xlabel("Document Category")
plt.ylabel("Number of Documents")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X = df_ml["cleaned_text"]
y = df_ml["category"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training documents:", len(X_train))
print("Testing documents:", len(X_test))

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

```python
%pip install imbalanced-learn
```

```python
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import RandomOverSampler

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

balanced_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            max_features=5000,
            ngram_range=(1, 2),
            min_df=1,
            sublinear_tf=True
        )
    ),
    (
        "oversampler",
        RandomOverSampler(random_state=42)
    ),
    (
        "classifier",
        LinearSVC(random_state=42)
    )
])

balanced_model.fit(X_train, y_train)

print("Model training completed!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

y_pred = balanced_model.predict(X_test)

print(
    "Accuracy:",
    round(accuracy_score(y_test, y_pred) * 100, 2),
    "%"
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        labels=["invoice", "bills", "receipt"],
        zero_division=0
    )
)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=["invoice", "bills", "receipt"],
    xticks_rotation=45
)

plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
import joblib

MODEL_PATH = BASE_DIR / "document_classifier.pkl"

joblib.dump(balanced_model, MODEL_PATH)

print("Model saved successfully:")
print(MODEL_PATH)
```

```python
sample_text = """
INVOICE
Invoice Number: INV-1001
Customer Name: ABC Company
Subtotal: 5000
Tax: 500
Total Amount: 5500
Payment Due Date: 20 October 2026
"""

sample_cleaned = clean_text(sample_text)

prediction = balanced_model.predict([sample_cleaned])[0]

print("Predicted category:", prediction)
```
